```python
import numpy as np
import matplotlib.pyplot as plt
```

```python

meses=2
p=int(meses*4*7/2)
print(type(p))
pool=407
values=6
d = []
s_d = []
for i in range(values+1):
    d.append((407-values)/(407-i))
    s_d.append((values-i)/(407-i))
A=np.zeros((len(d),len(d)))
for i in range(len(d)-1):
    A[i,i]=d[i]
    A[i,i+1]=s_d[i]
A[-1,-1]=d[-1]
B=np.linalg.matrix_power(A,p)
x=np.linspace(0,values,values+1)
y=B[0,:]
```

```python
plt.plot(x,y)
plt.show()
```
